#Task 1

```python
%%sql
CREATE TABLE NBA_Merch (
    product_id INTEGER PRIMARY KEY,
    name TEXT,
    price REAL
) STRICT;

INSERT INTO NBA_Merch (name, price)
VALUES
    ('Superbowl Jersey', 120.00),
    ('Browns Hat', 75.00),
    ('Steelers Key Chain', 35.00),
    ('Ravens Hoodie', 150.00);
```

#Task 2

```python
%%sql
/* Show all products that cost more than $50 */
SELECT * FROM NBA_Merch WHERE price > 50;
```

#Task 3

```python
%%sql
/* Update a price */
UPDATE NBA_Merch SET price = 99.99 WHERE name = 'Superbowl Jersey';
/* Delete an item */
DELETE FROM NBA_Merch WHERE product_id = 1;
```
