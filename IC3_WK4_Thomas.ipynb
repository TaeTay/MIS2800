```python
#Block 1
#Instructions: Create a table named stock_performance with columns for ticker, price_before_game, price_week_after
%%sql
CREATE TABLE IF NOT EXISTS stock_performance(
    ticker TEXT,
    price_before_game REAL,
    price_week_after REAL,
    trading_volume_millions INTEGER
) STRICT;
```

```python
#Block 3
#Instructions: Create a table named Ad_Buys with columns for ticker, company_name, ad_cost_millions, airtime_second
#The ticker should be a TEXT data type and serve as the Primary Key
#Select an appropriate data_type for the other columns
#Remember: use the STRICT keyword to ensure the datatype is enforced
%%sql
CREATE TABLE Ad_Buys (
  ticker TEXT PRIMARY KEY,
  company_name TEXT,
  ad_cost_millions REAL,
  airtime_seconds INTEGER
) STRICT;
```

```python
#Block 5
#Instructions: Show all the columns from the stock_performance table
%%sql
SELECT * FROM stock_performance;
```

```python
#Block 6
#Instructions: Show the ticker and price before the game
#Sort the ticker in alphabetical order from the stock_performance table
%%sql
SELECT ticker, price_before_game FROM stock_performance ORDER BY ticker ASC;
```

```python
#Block 7
#Instructions: Show all the columns for brands that have a ticker beginning with the letter T in the stock_performance table
%%sql
SELECT ticker FROM stock_performance WHERE ticker LIKE 'T%';
```

```python
#Block 8
#Instructions: Show only the ticker and price after game from the stock performance table if the price is above 150
%%sql
SELECT ticker, price_week_after FROM stock_performance WHERE price_week_after > 150;
```

```python
#Block 9
#Instructions: Show all the columns from the ad buys table
%%sql
SELECT * FROM Ad_Buys;
```

```python
#Block 10
#Instructions: Show the ticker and company name from the ad buys table in descending order by company name
%%sql
SELECT ticker, company_name FROM Ad_Buys ORDER BY company_name DESC;
```

```python
#Block 11
#Instructions: Show the ticker and airtime seconds from the ad buys table. Filter to show only if the ticker begins with a T and airtime seconds is 0
%%sql
SELECT ticker, airtime_seconds FROM Ad_Buys WHERE ticker LIKE 'T%' AND airtime_seconds = 0;
```
